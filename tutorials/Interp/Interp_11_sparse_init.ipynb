# Interpretability 11: sparse initialization

```python
import torch
from xkan import KANNetwork
from xkan.viz import plot

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(device)

model = KANNetwork([5,5,5,1], sparse_init=False, device=device)
x = torch.rand(100,5).to(device)
plot(model, x);
```
