# API 3: Extracting activation functions

The KAN diagrams give intuitive illustration, but sometimes we may also want to extract the values of activation functions for more quantitative tasks. Using the indexing convention introduced in the indexing notebook, each edge is indexed as $(l,i,j)$, where $l$ is the layer index, $i$ is the input neuron index, and $j$ is output neuron index.

```python
import torch
import matplotlib.pyplot as plt
from xkan import KANNetwork
from xkan.interp import edge_range
from xkan.viz import plot, plot_edge

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(device)

# create a KAN: 2D inputs, 1D output, and 5 hidden neurons. cubic spline (k=3), 3 grid intervals (grid=3).
model = KANNetwork([2,5,1], grid=3, k=3, seed=1, device=device)
x = torch.normal(0,1,size=(100,2)).to(device)
plot(model, x, beta=100);
```

```python
l = 0
i = 0
j = 3
# the edge's inputs (sorted) and outputs over x
x_edge, y_edge = plot_edge(model, x, l, i, j)
```

`edge_range` gives the range of an activation function over the data: (x min, x max, y min, y max).

```python
edge_range(model, x, l, i, j)
```
