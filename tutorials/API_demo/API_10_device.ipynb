# Demo 10: Device

All other demos have by default used device = 'cpu'. In case we want to use cuda, we should pass the device argument to model and dataset.

```python
import time
import torch
from xkan import KANNetwork, create_dataset, fit


device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(device)

#device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
device = 'cpu'
print(device)
```

```python
model = KANNetwork([4,100,100,100,1], grid=3, k=3, seed=0).to(device)
f = lambda x: torch.exp((torch.sin(torch.pi*(x[:,[0]]**2+x[:,[1]]**2))+torch.sin(torch.pi*(x[:,[2]]**2+x[:,[3]]**2)))/2)
train, test = create_dataset(f, n_var=4, train_num=1000, device=device)

# train the model
#history = fit(model, train, test, steps=20, lamb=1e-3, lamb_entropy=2.)
start = time.time()
history = fit(model, train, test, opt="adam", lr=1e-3, steps=50, lamb=1e-3, lamb_entropy=5., update_grid=False)
print(f'{time.time() - start:.1f} s', 'train RMSE', history['train/y'][-1] ** 0.5, 'test RMSE', history['val/y'][-1] ** 0.5)
```
